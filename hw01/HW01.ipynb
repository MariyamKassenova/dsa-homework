```python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
## [База]
#1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
#2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
#3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
#4. Сколько заказов не оплачено и на какую сумму?
#5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.
```

```python
# 1
cities_clean = [city.strip().capitalize() for city in cities]
print(cities_clean)
```

```python
#2 Преобразуйте `amounts` из строк в числа: список `amounts_int`.
amounts_int = [int(amount) for amount in amounts]
print(amounts_int)
```

```python
#3 Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
orders = len(amounts_int)
total_sum = sum(amounts_int)
avarage_check = round(total_sum/orders)
print("Общее число заказов", orders)
print("Сумма заказов", total_sum)
print("Средний чек", avarage_check)
```

```python
#4 Сколько заказов не оплачено и на какую сумму?
unpaid_count = paid.count(False)
unpaid_sum = sum(amount for amount, is_paid in zip(amounts_int,paid)if not is_paid)
print("Количество отказов", unpaid_count)
print("Сумма отказов", unpaid_sum)
```

```python
#5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.
for city, amount, status in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {'Оплачен' if status else 'Не оплачен'}")
```

```python

## [Стандарт]
#6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
#7. Номер (индекс) и город самого крупного **оплаченного** заказа.
#8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
#9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?
```

```python
#6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
almaty_sum = 0
astana_sum = 0
for city, amount, status in zip(cities_clean, amounts_int, paid):
    if status and city == "Almaty":
        almaty_sum += amount
    elif status and city == "Astana":
        astana_sum += amount
print("Выручка Алматы:", almaty_sum)
print("Выручка Астаны:", astana_sum)
```

```python

#7. Номер (индекс) и город самого крупного **оплаченного** заказа.
max_index = 0
max_city = ""
max_amount = 0

for i, (city, amount, status) in enumerate(zip(cities_clean, amounts_int, paid)):
    if status and amount > max_amount:
        max_index = i
        max_city = city
        max_amount = amount

print("Индекс:", max_index)
print("Город:", max_city)
print("Сумма:", max_amount)
```

```python
#8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
weekend_orders = 0
for day, status in zip(days, paid):
  if status and day == "сб" or day == "вс":
    weekend_orders += 1
weekend_percent = (weekend_orders/len(days))*100
print("Доля заказов:", weekend_orders)
print("Процент:", round(weekend_percent))
```

```python
#9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит.
#А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?
months = 0
savings = 0
while savings < 400000:
  savings += 15000
  months += 1
print("Через месяцев:", months)
print("Копилка:", savings)
```
